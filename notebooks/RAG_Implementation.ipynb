# RAG Implementation

This notebook demonstrates the full Retrieval-Augmented Generation (RAG) pipeline using LangChain and ChromaDB.

```python
import sys
sys.path.insert(0, '../src')

from dotenv import load_dotenv
load_dotenv('../.env')
```

## 1. Load Documents

```python
from ingestion import load_documents

documents = load_documents('../data/sample')
print(f'Loaded {len(documents)} pages')
```

## 2. Split into Chunks

```python
from chunking import split_documents

chunks = split_documents(documents)
print(f'Split into {len(chunks)} chunks')
```

## 3. Create Embeddings

```python
from embeddings import create_embeddings

embeddings = create_embeddings()
```

## 4. Build Vector Store

```python
from vector_store import create_vector_store

vector_store = create_vector_store(chunks, embeddings)
print(f'Vector store has {vector_store._collection.count()} entries')
```

## 5. Create Retriever

```python
from retrieval import create_retriever

retriever = create_retriever(vector_store)
```

## 6. Test Retrieval

```python
results = retriever.invoke('What is the transformer architecture?')
print(f'Retrieved {len(results)} documents')
print()
for i, doc in enumerate(results, 1):
    source = doc.metadata.get('source_file', 'unknown')
    print(f'{i}. [{source}]')
    print(f'   {doc.page_content[:200]}...')
    print()
```

## 7. Set Up LLM and RAG Chain

```python
from generation import create_llm
from rag_chain import create_rag_chain

llm = create_llm()
rag_chain = create_rag_chain(retriever, llm)
```

## 8. Ask a Question

```python
result = rag_chain('What is the transformer architecture?')
print('Answer:')
print(result['answer'])
print()
print('Sources:')
for doc in result['sources']:
    print(f'  - {doc.metadata.get("source_file", "unknown")}')
```
